```
from pathlib import Path
from importlib.machinery import SourceFileLoader
ROOT = None
for candidate in [Path.cwd(), *Path.cwd().parents]:
    helper = candidate / 'PUBLIC' / 'handoff' / 'PUBLIC_18_business_storyline_260520' / '18_business_storyline_helper.py'
    if helper.exists():
        ROOT = candidate
        break
if ROOT is None:
    raise FileNotFoundError('helper not found')
mod = SourceFileLoader('storyline18', str(ROOT / 'PUBLIC' / 'handoff' / 'PUBLIC_18_business_storyline_260520' / '18_business_storyline_helper.py')).load_module()
summary = mod.run(finalize=False)
summary
```
